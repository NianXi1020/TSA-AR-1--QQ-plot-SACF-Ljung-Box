```python
# Part (a): Simulate AR(1), fit model, and plot residuals
from IPython.display import Image, display
from ar1_analysis_utils import (
    compute_residuals,
    fit_ar1,
    plot_residuals,
    simulate_ar1,
    standardize,
)

series = simulate_ar1(n=30, phi=0.5, sigma=1.0, seed=123)
intercept_hat, phi_hat = fit_ar1(series)
residuals = compute_residuals(series, intercept_hat, phi_hat)
standardized_residuals, residual_std = standardize(residuals)
plot_residuals(residuals, 'residuals_plot.png')
display(Image(filename='residuals_plot.png'))
print(f'Estimated intercept: {intercept_hat:.4f}')
print(f'Estimated phi: {phi_hat:.4f}')

```

```python
# Part (b): Display normal QQ plot of standardized residuals
from ar1_analysis_utils import normal_quantiles, plot_qq

theoretical_quantiles = normal_quantiles(len(standardized_residuals))
empirical_quantiles = sorted(standardized_residuals)
plot_qq(theoretical_quantiles, empirical_quantiles, 'qq_plot_standardized_residuals.png')
display(Image(filename='qq_plot_standardized_residuals.png'))

```

```python
# Part (c): Display sample ACF of residuals
import math
from ar1_analysis_utils import plot_acf, sample_acf

acf_values = sample_acf(residuals, 10)
conf_limit = 1.96 / math.sqrt(len(residuals))
plot_acf(acf_values, 'residuals_acf.png', conf_limit)
display(Image(filename='residuals_acf.png'))

```

```python
# Part (d): Calculate Ljung-Box statistic up to lag 8
from ar1_analysis_utils import ljung_box

ljung_box_stat, ljung_box_p = ljung_box(residuals, 8)
print(f'Ljung-Box statistic (lag 8): {ljung_box_stat:.4f}')
print(f'P-value: {ljung_box_p:.4f}')

```
